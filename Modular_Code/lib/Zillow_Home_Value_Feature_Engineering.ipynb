### **Handling Missing Values**

Handling missing values is an important step in data preprocessing. Missing data can lead to biased results or affect the performance of machine learning models. There are several ways to handle missing values such as:

* Dropping missing values: This approach involves dropping the rows or columns that contain missing values. However, this method may lead to loss of information, especially if a large number of rows or columns have missing values.

* Imputing missing values: This approach involves filling in the missing values with a value that is either derived from other observations or estimated using statistical methods. Common imputation methods include mean imputation, median imputation, and regression imputation.

* Using advanced methods: There are more advanced methods for handling missing data such as multiple imputation, k-nearest neighbors imputation, and matrix completion. These methods can be more accurate than simple imputation methods, but they may also be more computationally intensive.

In general, the choice of how to handle missing values will depend on the specific problem and dataset. It is important to carefully consider the impact of missing values on the analysis and choose an appropriate method that balances the need for accurate results with the computational complexity of the method.





```python
mis_val = [var for var in df.columns if df[var].isnull().sum()>0]
df[mis_val].isnull().sum()
```

### **Dropping columns having >60% missing values:**

```python
df.shape
```

```python
def drop_columns_with_max_missing_values(df):
    """
    Drops columns from the DataFrame that have more than 60% missing values.

    Parameters:
    df (DataFrame): The DataFrame containing the data.

    Returns:
    DataFrame: The DataFrame with columns containing more than 60% missing values dropped.
    """
    # Identify columns with missing values
    mis_var = [var for var in df.columns if df[var].isnull().sum() > 0]

    # Print the count of missing values for each identified column
    print(df[mis_var].isnull().sum())

    # Set the limit for the maximum allowable missing values (60%)
    limit = np.abs((df.shape[0] * 0.6))

    # Identify columns with more than 60% missing values
    var_to_be_dropped = [var for var in mis_var if df[var].isnull().sum() > limit]

    # Print the columns with more than 60% missing values
    print('Columns with more than 60% missing values: \n\n', var_to_be_dropped)
    print('\n##### Dropping columns with more than 60% missing values #####\n')

    # Drop columns with more than 60% missing values
    df.drop(columns=var_to_be_dropped, axis=1, inplace=True)

    # Print the remaining columns after dropping
    print('Remaining columns: \n', df.columns)

    # Return the DataFrame with dropped columns
    return df

# Call the function to drop columns with more than 60% missing values
df = drop_columns_with_max_missing_values(df)
```

### **Encoding categorical variables:**

Encoding categorical variables is a crucial preprocessing step in data analysis and machine learning tasks. Categorical variables represent qualitative data with discrete categories or levels, such as gender, color, or country. However, most machine learning algorithms require numerical inputs, making it necessary to convert categorical variables into a numerical format. Encoding categorical variables involves assigning a unique numerical value to each category, allowing algorithms to understand and process the data effectively.

There are various encoding techniques available, including one-hot encoding, label encoding, and target encoding.

One-hot encoding creates binary columns for each category, where a value of 1 indicates the presence of the category and 0 indicates absence.

Label encoding assigns a unique integer to each category, while target encoding replaces categories with the mean of the target variable for each category.

Choosing the appropriate encoding technique depends on the nature of the data and the requirements of the machine learning algorithm.

```python
cat_vars = [var for var in df.columns if df[var].dtypes=='O']
cat_vars
```

```python
from sklearn.preprocessing import LabelEncoder

def encode_categorical_variables(df, cat_vars):
    """
    Performs Label Encoding on categorical variables in the DataFrame.

    Parameters:
    df (DataFrame): The DataFrame containing the data.
    cat_vars (list): A list of categorical variables to encode.

    Returns:
    DataFrame: The DataFrame with categorical variables encoded.
    """
    # Print information about the encoding process
    print('##### Performing Label Encoding on Categorical Variables #####')
    print('Categorical variables: ', cat_vars)

    # Iterate over each categorical variable
    for i in range(len(cat_vars)):
        var = cat_vars[i]

        # Initialize a LabelEncoder object
        var_le = LabelEncoder()

        # Encode the categories of the variable and store the encoded labels
        var_labels = var_le.fit_transform(df[var])

        # Create a dictionary to map encoded labels to their original categories
        var_mappings = {index: label for index, label in enumerate(var_le.classes_)}

        # Create a new column with the encoded labels and drop the original categorical variable
        df[(var + '_labels')] = var_labels
        df.drop(columns=var, axis=1, inplace=True)

    # Return the DataFrame with categorical variables encoded
    return df

# Call the function to encode categorical variables in the DataFrame
df = encode_categorical_variables(df, cat_vars)
```

### **Checking & Removing Outliers:**


In regression analysis, the presence of outliers can significantly impact the accuracy and reliability of the model's predictions. Outliers are data points that deviate markedly from the rest of the dataset and can exert disproportionate influence on the regression coefficients and overall model fit. When outliers are included in the analysis, they can bias the estimated coefficients, leading to misleading interpretations of the relationships between the independent and dependent variables. Outliers may also increase the variability of the error term, violating the assumption of homoscedasticity, and undermine the statistical inference derived from the regression results. Therefore, removing outliers is essential in regression analysis to improve the robustness of the model, enhance the accuracy of parameter estimates, and ensure the validity of the statistical inferences drawn from the analysis. By identifying and excluding outliers, regression models can better capture the underlying patterns and relationships in the data, resulting in more reliable predictions and actionable insights for decision-making purposes.



 In the context of outlier removal, the Z-score is particularly useful as it provides a systematic and objective way to identify extreme values that fall far from the mean of the dataset. Typically, data points with a Z-score greater than a certain threshold, often set at 2 or 3 standard deviations from the mean, are considered outliers and can be removed from the dataset.

#### **Correlation**

Correlation coefficient is used to measure the strength of relationship between two variables. It indicates that as the value of one variable changes the other variable changes in a specific direction with some magnitude. There are various ways to find correlation between two variables, one of which is Pearson correlation coefficient.
It measures the linear relationship between two continuous variables.

Let's say $x$ and $y$ are two continuous variables, the Pearson correlation coefficient between them can be found by the following formula.


$$r = \frac{ \sum_{i=1}^{n}(x_i-\bar{x})(y_i-\bar{y}) }{%
        \sqrt{\sum_{i=1}^{n}(x_i-\bar{x})^2}\sqrt{\sum_{i=1}^{n}(y_i-\bar{y})^2}}$$

where $x_i$ and $y_i$ represents the $i^{th}$ value of the variables.
The value of $r$ ranges between $-1$ and $+1$.

Their strength of relationship is measured by the absolute value of coefficient, whereas the sign of the coefficient indicates the direction of the relationship.

### **Checking for Multi-Colinearity:**


Multicollinearity refers to the phenomenon where two or more independent variables in a regression model are highly correlated with each other. This correlation can cause issues in the model estimation process, leading to unreliable coefficient estimates and inflated standard errors. Multicollinearity does not affect the predictive power of the model but rather impacts the interpretation of individual coefficients. When multicollinearity is present, it becomes challenging to isolate the effect of each independent variable on the dependent variable accurately. High multicollinearity can also make it difficult to identify the relative importance of different predictors in explaining the variation in the target variable.
